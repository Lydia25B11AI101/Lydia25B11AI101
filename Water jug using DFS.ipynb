```python
def jug(a, b, t):
  seen = set()
  def go(s, path):
    x, y = s
    if s in seen:
      return None
    seen.add(s)
    path = path + (s,)
    if t in s:
      return path
    p, q = min(x, b-y), min(y, a-x)
    for n in [(a, y), (x, b), (0, y), (x, 0), (x-p, y+p), (x+q, y-q)]:
      r = go(n, path)
      if r:
        return r
    return None
  return go((0, 0), ())

print(jug(4, 3, 2))
```
